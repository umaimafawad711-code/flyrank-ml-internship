## 1. My rule and its reason codes
Rule in plain words: For Lane 4 (CTR / Engagement Opportunity Scoring), prioritize content pages that have high visibility (impressions) but a lagging click-through rate (CTR) or suboptimal average ranking position. These represent immediate optimization opportunities where minor title, meta, or snippet adjustments can capture missing traffic.

Reason Codes:

HIGH_IMPRESSION_CTR_GAP: High impression volume combined with lower-than-expected CTR.

TOP_POSITION_STAGNATION: Ranking in top position tiers but failing to capture optimal traffic share.

VOLUME_OPPORTUNITY: Strong trailing query volume suitable for quick-win optimization.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
from datasets import load_dataset
from google.colab import userdata

# Load warehouse dataset securely using HF_TOKEN
hf_token = userdata.get('HF_TOKEN')
dataset = load_dataset("FlyRank/internship-warehouse", 'fact_content_query_90d', token=hf_token)
df = pd.DataFrame(dataset['train'].select(range(5000)))

print(f"Dataset loaded successfully with {len(df)} rows for rule processing.")
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*
This section scores the dataset using our baseline heuristic, assigns reason codes and action labels, sorts the queue in descending order of priority, and exports the final output to work/outputs/baseline_action_score.csv.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd

# Ensure output directory exists
os.makedirs('work/outputs', exist_ok=True)

# Code the score based on impressions and position gaps
df['score'] = df['impressions_90d'] / (df['avg_position_90d'] + 1)
df['reason_code'] = 'HIGH_IMPRESSION_CTR_GAP'
df['action_label'] = 'OPTIMIZE_META_AND_SNIPPET'

# Rank everything and select required columns for output queue
ranked_queue = df[['content_hash_id', 'query_hash_id', 'score', 'reason_code', 'action_label']].sort_values(by='score', ascending=False)

# Write to CSV
output_path = 'work/outputs/baseline_action_score.csv'
ranked_queue.to_csv(output_path, index=False)
print(f"Ranked queue successfully written to {output_path} containing {len(ranked_queue)} rows.")
```
