## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load dataset directly from the raw GitHub URL
url = "https://raw.githubusercontent.com/umaimafawad711-code/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Filter for pages with meaningful visibility (impressions >= 100)
visible = df[df["impressions_90d"] >= 100]

# 1. Calculate mean CTR by position tier
ctr_by_position = visible.groupby("position_tier")["ctr"].mean().round(4)

# 2. Count top-rank pages (Position 1 to 10) that have a low CTR (< 1.0%)
top_rank_low_ctr = visible[(visible["avg_position"] > 0) & (visible["avg_position"] <= 10) & (visible["ctr"] < 1.0)]

print("--- REAL NUMBERS FOR LANE 4 ---")
print("\n1. Mean CTR by Position Tier (Impressions >= 100):")
print(ctr_by_position.to_string())

print(f"\n2. Total visible pages analyzed: {len(visible)}")
print(f"3. Top-rank pages (Position 1-10) with low CTR (< 1%): {len(top_rank_low_ctr)}")
```
