## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Machine learning / multi-signal scoring is required because CTR behavior is tangled with varying impressions, position tiers, and engagement rates, making a data-driven score much more adaptive than a rigid rule
